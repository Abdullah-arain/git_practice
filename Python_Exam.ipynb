```python
# 1. Interplanetary Weight Calculator
# Extend the planetary weight calculator to include options for multiple planets. Write a Python program that prompts a user for their weight on Earth and allows them to select a planet from the following list:

# Mars (37.8% of Earth weight)
# Moon (16.5% of Earth weight)
# Jupiter (252% of Earth weight)
# The program should calculate and display the weight accordingly.

user_input = int(input("Enter your weight on Earth"))
select = int(input("""
Select any of one from given:
1. Mars
2. Jupiter
3. Moon
"""))

if select == 1 or select == "mars":
  print(f"Your weight on Mars would be: {user_input * 0.378:.2f}")
elif select == 2 or select == "jupiter":
  print(f"Your weight on Jupiter would be: {user_input * 2.52:.2f}")
elif select == 3 or select == "moon":
  print(f"Your weight on the Moon would be: {user_input * 0.165:.2f}")
else:
  print("Something went wrong")
```
